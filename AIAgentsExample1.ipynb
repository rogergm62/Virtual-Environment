````python
import os

from dotenv import load_dotenv
from litellm import completion

load_dotenv()

api_key = os.getenv("OPENAI_API_KEY")

if not api_key:
    raise ValueError("OPENAI_API_KEY was not found.")

os.environ["OPENAI_API_KEY"] = api_key.strip()

def generate_response(messages: list[dict]) -> str:
    response = completion(
        model="openai/gpt-4o",
        messages=messages,
        max_tokens=1024
    )

    return response.choices[0].message.content

messages = [
    {
        "role": "system",
        "content": "You are an expert software engineer."
    },
    {
        "role": "user",
        "content": "Write a script in Python able to draw a simple quadratic function."
    }]

response = generate_response(messages)

print(response)
````

<p style="font-size:28px;"> What does this Python script do?.</p>
<p style="font-size:20px;">Here is the same script with added comments describing its job: </p>

````python
import os  # Provides access to environment variables and operating system features.

from dotenv import load_dotenv  # Loads variables from a .env file.
from litellm import completion  # Imports LiteLLM's completion function.

load_dotenv()  # Loads environment variables from the .env file.

api_key = os.getenv("OPENAI_API_KEY")  # Gets the OpenAI API key from the environment.

if not api_key:  # Checks whether the API key was found.
    raise ValueError("OPENAI_API_KEY was not found.")  # Stops the program if the key is missing.

os.environ["OPENAI_API_KEY"] = api_key.strip()  # Removes extra spaces and saves the key back into the environment.


def generate_response(messages: list[dict]) -> str:  # Defines a function that sends messages to the AI model.
    response = completion(  # Sends a request to the selected AI model.
        model="openai/gpt-4o",  # Specifies the OpenAI GPT-4o model.
        messages=messages,  # Sends the conversation messages to the model.
        max_tokens=1024  # Limits the maximum number of tokens in the response.
    )

    return response.choices[0].message.content  # Returns the text generated by the model.

messages = [  # Creates the list of messages that will be sent to the AI.
    {
        "role": "system",  # Defines instructions that control how the AI should behave.
        "content": "You are an expert software engineer."  # Tells the AI to act as a software engineer.
    },
    {
        "role": "user",  # Identifies this message as coming from the user.
        "content": "Write a script in Python able to draw a simple quadratic function."  # Gives the AI the task.
    }
]

response = generate_response(messages)  # Calls the function and stores the AI's response.

print(response)  # Displays the AI-generated response in the terminal.
````

````python
To draw a simple quadratic function in Python, we can use libraries like `matplotlib` for plotting charts and `numpy` for numerical operations. A quadratic function can be represented in the form \( f(x) = ax^2 + bx + c \), where \( a \), \( b \), and \( c \) are constants. Here's a step-by-step guide to create a Python script that plots a quadratic function:

1. Install necessary libraries if you haven't already:
   ```bash
   pip install matplotlib numpy
   ```

2. Create the Python script to plot the quadratic function:

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_quadratic(a, b, c):
    # Create an array of x values from -10 to 10 with a 0.1 step
    x_values = np.arange(-10, 10, 0.1)
    # Compute the corresponding y values for the quadratic function
    y_values = a * x_values**2 + b * x_values + c

    # Plot the function
    plt.figure(figsize=(8, 6))
    plt.plot(x_values, y_values, label=f'{a}x² + {b}x + {c}')
    plt.title('Plot of the Quadratic Function')
    plt.xlabel('x')
    plt.ylabel('f(x)')
    plt.axhline(0, color='black', linewidth=0.8, linestyle='--')  # Add x-axis
    plt.axvline(0, color='black', linewidth=0.8, linestyle='--')  # Add y-axis
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.legend()
    plt.show()

# Example usage: plot the function f(x) = 1x² + 0x + 0
plot_quadratic(a=1, b=0, c=0)
```

### Explanation:

- **Imports and Setup**: We import `numpy` for easy numerical calculations and `matplotlib.pyplot` for plotting.
- **Function Definition**: We define `plot_quadratic(a, b, c)` which takes the coefficients of the quadratic equation as arguments.
- **Generating Data**: We generate a range of x values using `np.arange()`. Corresponding y values are computed based on the quadratic formula.
- **Plotting**: We use `plt.plot()` to create a 2D line plot of the data. We also add grid lines, a legend, and axis lines for better visualization.
- **Display**: Finally, `plt.show()` displays the plot in a window.

You can modify the coefficients when calling `plot_quadratic()` to visualize different quadratic functions. For example, `plot_quadratic(1, 2, 1)` will plot \( f(x) = x^2 + 2x + 1 \).
````
