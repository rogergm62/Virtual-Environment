<h2 style="color:#1474c4; font-size:30px;">
Why use a virtual environment?
</h2>

<p style="font-size:20px;"> Keeps your projects isolated and independent.</p>

<p style="font-size:20px;"> Avoids dependency conflicts between projects.</p>

<p style="font-size:20px;"> Makes it easier to share and reproduce your environment.</p>

<p style="font-size:20px;">
<code>uv</code> is fast, simple, and designed for modern Python development.
</p>

<div style="margin-top:30px; padding:18px 20px; background-color:#eef6ff; border-left:6px solid #2583d8; border-radius:4px; font-size:19px;">
<strong style="color:#1474c4;">Tip:</strong>
Think of a virtual environment as your own <em>workspace</em>, where your project can live without interfering with other Python projects.
</div>

<div style="margin-top:20px; border:2px solid #2583d8; border-radius:10px; background-color:#ffffff; padding:25px 30px; font-family:Arial, sans-serif;">

<h2 style="
     color:#1474c4;
     font-size:28px;
     margin-top:0;">
        Step 1: Install uv
</h2>

<p style="
    font-size:19px;
    line-height:1.6;">
Before creating a virtual environment, download and install
<code>uv</code> using the official Astral installer.
</p>

<p style="
    font-size:19px;
    line-height:1.6;">
Open <strong>PowerShell</strong> and run:
</p>
</div>


```python
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"
```

<div style="margin-top:20px; border:2px solid #2583d8; border-radius:10px; background-color:#ffffff; padding:25px 30px; font-family:Arial, sans-serif;">

<h2 style="
     color:#1474c4;
     font-size:28px;
     margin-top:0;">
        Step 2: Install Python
</h2>

<p style="
    font-size:19px;
    line-height:1.6;">
Until now (2026), uv runs smoothly in <code>Python 3.12</code> Make sure you have that version installed
</p>

<p style="
    font-size:19px;
    line-height:1.6;">
Open <strong>PowerShell</strong> and run:
</p>
</div>


```python
py --list
```

<div style="margin-top:20px; border:2px solid #2583d8; border-radius:10px; background-color:#ffffff; padding:25px 30px; font-family:Arial, sans-serif;">

<h2 style="
     color:#1474c4;
     font-size:28px;
     margin-top:0;">
        Step 3: Create your Project Folder.
</h2>

<p style="
    font-size:19px;
    line-height:1.6;">
Creating a folder per project and a virtual environment inside that folder is the correct, professional, and future‑proof way to work with Python.
</p>

<p style="
    font-size:19px;
    line-height:1.6;">
Open <strong>PowerShell</strong> and run:
</p>
</div>

```python
mkdir ProjectFolder
```

<div style="margin-top:20px; border:2px solid #2583d8; border-radius:10px; background-color:#ffffff; padding:25px 30px; font-family:Arial, sans-serif;">

<h2 style="
     color:#1474c4;
     font-size:28px;
     margin-top:0;">
        Step 4: Tell uv to use Python 3.12 explicitly.
</h2>

<p style="
    font-size:19px;
    line-height:1.6;">
Inside this folder, we need to force uv to create a .venv using Python 3.12. Installing Python 3.12 will not cause conflicts, but it will make your development environment more flexible
</p>

<p style="
    font-size:19px;
    line-height:1.6;">
Open <strong>PowerShell</strong> and run:
</p>
</div>

```python
uv python pin 3.12
```

<div style="margin-top:20px; border:2px solid #2583d8; border-radius:10px; background-color:#ffffff; padding:25px 30px; font-family:Arial, sans-serif;">

<h2 style="
     color:#1474c4;
     font-size:28px;
     margin-top:0;">
        Step 5: Start UV.
</h2>

<p style="
    font-size:19px;
    line-height:1.6;">
This will automatically:
    
create a pyproject.toml

create a .venv folder

create a uv.lock

set up a clean Python project structure
You don’t need to manually create anything.
</p>

<p style="
    font-size:19px;
    line-height:1.6;">
Open <strong>PowerShell</strong> and run:
</p>
</div>

```python
uv init
```

<div style="margin-top:20px; border:2px solid #2583d8; border-radius:10px; background-color:#ffffff; padding:25px 30px; font-family:Arial, sans-serif;">

<h2 style="
     color:#1474c4;
     font-size:28px;
     margin-top:0;">
        Step 6: Add whatever libraries you need.
</h2>

<p style="
    font-size:19px;
    line-height:1.6;">
uv will:

update your pyproject.toml

update your uv.lock

install the library into your .venv

make the import work immediately
</p>

<p style="
    font-size:19px;
    line-height:1.6;">
Open <strong>PowerShell</strong> and run (example):
</p>
</div>

```python
uv add openai
uv add notebook
uv add gradio
uv add numpy
uv add pandas
uv add fastapi
```

<div style="margin-top:20px; border:2px solid #2583d8; border-radius:10px; background-color:#ffffff; padding:25px 30px; font-family:Arial, sans-serif;">

<h2 style="
     color:#1474c4;
     font-size:28px;
     margin-top:0;">
        Step 7: Run UV Sync (Optional).
</h2>

<p style="
    font-size:19px;
    line-height:1.6;">
uv will:

Read your pyproject.toml

Create a .venv folder automatically

Install all dependencies listed

Lock everything in uv.lock. This ensures reproducibility — same versions every time.

Once a project is running, running uv sync again is harmless; it just ensures everything is consistent.

uv add already performs a sync. But you can run it if you want to:

reinstall everything cleanly

verify the environment

ensure reproducibility

It won’t break anything.
</p>

<p style="
    font-size:19px;
    line-height:1.6;">
Open <strong>PowerShell</strong> and run (example):
</p>
</div>

```python
uv sync
```
