<h2 style="color:#1474c4; font-size:30px;">
Decorators.
</h2>

<p style="font-size:20px;"> A decorator in Python is a function that adds extra behavior to another function</p>

<p style="font-size:20px;"> without changing the original function’s code directly.</p>

<p style="font-size:20px;"> So the decorator takes the original function, wraps extra behavior around it, and returns the new version.</p>

<p style="font-size:20px;"> They are commonly used for logging, authentication, timing, error handling, caching, and access control.</p>

<p style="font-size:20px;"> Decorator version: demonstrates an important Python programming technique and gives you evidence about how decorators can add reusable behavior to functions.</p>

```python
tools = {}

def register_tool(): # Registra las funciones como herramientas
    def herramientas(function):   # Store the function in the tools dictionary
        tools[function.__name__] = function
        return function
    return herramientas

# TOOL 1
@register_tool()
def Regular_greetings(name): # Give a regular greeting
    return f"Hello, {name}!"

# TOOL 2
@register_tool()
def Birthday_greetings(name): # Give a birthday greeting
    return f"Happy birthday, {name}!"

# ------------------------------------------------------------
# AGENT
# ------------------------------------------------------------

def greetings_agent(user_request):
    request = user_request.lower()
    if "regular" in request:
        name = input("Give me you name for a regular greeting: ")
        # Call the tool from the registry
        result = tools["Regular_greetings"](name)
        return result

    elif "birthday" in request:
        name = input("Give me you name for a birthday greeting: ")
        # Call the tool from the registry
        result = tools["Birthday_greetings"](name)
        return result

    else:
        return (
            "I don't know how to do that. "
            "I can only give you regular or birthday greetings.")

# ------------------------------------------------------------
# MAIN PROGRAM
# ------------------------------------------------------------

print("Greetings Machine")
print("------------")
print("Available tools:")
for tool_name in tools:
    print("-", tool_name)
print()

while True:
    user_request = input("What would you like me to do? Type 'quit' to exit: ")
    if user_request.lower() == "quit":
        print("Goodbye!")
        break
    response = greetings_agent(user_request)
    print(response)
    print()
```
